```python
''' Angle between hour and minute hand is Given a string s representing time in 24-hour format "HH:MM",
compute the smallest angle in degrees between the hour and minute hands of an analog clock.'''
```

```python
n = input("Enter time(HH : MM)")

h , m = map(int, n.split(":"))
h %= 12

h_a = h * 30 + m * 0.5

m_a = m * 6

diff = abs(h_a - m_a)

ans = min(diff, 360 - diff)

print("Smallest angle in degree is : " , ans)
```

```python
# Write a program that will check whether the number is armstrong number or not.
```

```python
n = int(input("Enter a number: "))

original = n
digits = len(str(n))
total = 0

while n > 0:
    digit = n % 10
    total = total + (digit ** digits)
    n = n // 10

if total == original:
    print(original, "is an Armstrong no")
else:
    print(original, "is not an Armstrong no")
```

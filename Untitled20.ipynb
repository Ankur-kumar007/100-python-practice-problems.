```python
while True:
    print("/n======== Unit Convert ===========")
    print("1. cm to feet")
    print("2. km to miles")
    print("3. usd to inr")
    print("4. Exist")

    choise = int(input("Enter your choice (1-4): "))

    if choise == 1 :
        cm = float(input("Enter length in cm :"))
        ft = cm / 30.48
        print("Length in feet:" , ft)

    elif choise == 2 :
        km = float(input("Enter distance in km :"))
        miles = km * 0.631
        print("Distance in miles:", miles)

    elif choise == 3 :
        usd = float(input("Enter amount in usd:"))
        rate = 96.0
        inr = usd * rate
        print("Amount in INR :" , inr)
    elif choise == 4:
        print("Exist the program....")
        break
    else:
        print("Invalid choise! ")
```

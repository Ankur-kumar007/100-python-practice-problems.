```python
# Write a program that will give you the in hand salary after deduction of HRA(10%),DA(5%),PF(3%),
# and tax(if salary is between 5-10 lakh–10%),(11-20lakh–20%),(20< _ – 30%)(0-1lakh print k).
```

```python
# Take annual salary from user

salary = float(input("Enter your annual salary :"))

# Deduction

hra = salary * 0.10
da = salary * 0.05
pf = salary * 0.03

# Income tax calculation

if salary <= 100000:
    print("k")
    tax = 0

elif salary <= 500000 and salary <= 1000000:
    tax = salary * 0.10
elif salary <= 1100000 and salary <= 2000000:
    tax = salary * 0.20
elif salary < 2000000:
    tax = salary * 0.30
else:
    tax = 0.30

# In hand salary

in_hand_salary = salary - (hra + da + pf + tax)

# Display result

print("hra deduction :", hra)
print("da deduction :", da)
print("pf deduction :", pf)
print("tax deduction :", tax)
print("In_Hand_Salary :", in_hand_salary)
```

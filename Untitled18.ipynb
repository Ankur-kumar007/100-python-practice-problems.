```python
# Write a program that will take user input of (4 digits number) and check whether the number is narcissist number or not.
```

```python
n = int(input("Enter a four digit number :" ))

original = n
digits = len(str(n))
total = 0

while n>0:
    digits = n % 10
    total = total + (digits ** digits)
    n = n // 10

if total == original:
    print(original , "is a narcissist number")
else:
    print(original, "is not a narcissist number")

```

```python
# Write a program to find the volume of the cylinder. Also find the cost when ,when the cost of 1 litre milk is 40Rs.
```

```python
# Input coordinates
r = float(input("Enter the radius value :"))
h = float(input("Enter the height value :"))

# logic (calculate volumn)
v = 3.14 * r * r * h 
liter = v / 1000

# 1 liter 40 rupees
cost = liter * 40
# Print

print("Volume = ", v)
print("liter = ", liter)
print("cost = ", cost)
```

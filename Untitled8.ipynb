```python
# Write a program that take a user inputr of three angles and will find out whether it can form a triangle or not.
```

```python
a1 = int(input("Enter a1 angle: "))
a2 = int(input("Enter a2 angle: "))
a3 = int(input("Enter a3 angle: "))

if a1 > 0 and a2 > 0 and a3 > 0 and (a1 + a2 + a3 == 180):
    print("Yes, it can form a triangle")
else:
    print("No, it can't be form a triangle")
```

```python
# Write a program that will tell whether the given number is divisible by 3 & 6.
```

```python
num = int(input("Enter a number : "))

# check if it is divisible by 3 and 6
if num % 3 == 0 and num % 6 == 0:
    print ("Yes , it is divisible by both 3 and 6" , num)
else:
    print("No , it is not divisible by both 3 and 6" , num)
```
